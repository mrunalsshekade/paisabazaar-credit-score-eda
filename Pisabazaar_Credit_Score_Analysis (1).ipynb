# **Project Summary -**

This project presents an exploratory data analysis(EDA) of customer financial data sourced from Paisabazaar, one of India's leading financial marketplace platforms. The dataset contains 100,000 records across 28 features, capturing monthly financial behavior of customers including income, debt levels, payment patterns, credit history, and credit score categories.

The core business problem addressed in this project is credit risk assessment,identifying which customer behaviors and financial attributes most strongly influence whether a customer falls into a Good, Standard, or Poor credit score category. Accurate credit risk assessment is critical for financial institutions to minimize loan defaults, optimize lending decisions, and design appropriate financial products for different customer segments.

The analysis follows a structured exploratory approach using the UBM framework; Univariate, Bivariate, and Multivariate analysis across 20 meaningful charts. Before visualization, data wrangling was performed including removal of identifier columns, detection and handling of hidden irregular values, treatment of unrealistic age entries below 18, and imputation of missing values using statistically appropriate methods.

Key findings from the analysis include the identification of outstanding debt, payment delay frequency, and number of delayed payments as the strongest indicators of poor creditworthiness. A particularly notable discovery was that customers who pay only the minimum due amount consistently showed worse credit outcomes and higher payment delays than those who do not, challenging the conventional assumption that minimum payment compliance reflects responsible financial behavior.

The correlation analysis revealed that annual income and monthly inhand salary are perfectly correlated and effectively redundant features, while credit history age emerged as the strongest protective variable across the dataset, negatively correlated with nearly every risk indicator. Credit utilization ratio, often considered a key credit metric, was found to be a weak standalone predictor in this dataset, showing minimal variation across credit score categories.

The analysis also confirmed that income and occupation alone are insufficient predictors of creditworthiness. High income customers were present across all three credit score categories, and debt burden patterns were consistent across all occupations, indicating that behavioral and historical credit factors carry significantly more predictive weight than demographic or professional attributes.

These findings provide Paisabazaar with a clearer understanding of the behavioral and financial patterns that distinguish creditworthy customers from high risk ones. The insights generated through this analysis can directly support more informed lending decisions, improved customer segmentation, and better targeted financial product offerings across all customer groups.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
#creating a copy of dataset
df1 = df.copy()

#dropping unneessary columns
df1.drop(columns=['ID','Customer_ID','Name','SSN'], inplace=True)
````

````python
#checking irregular values
print(df1['Payment_of_Min_Amount'].unique())
````

````python
#replacing irregular value NM to NaN
df1['Payment_of_Min_Amount'] = df1['Payment_of_Min_Amount'].replace('NM',np.nan)

#checking missing values again
df1['Payment_of_Min_Amount'].isnull().sum()
````

````python
df1['Age'].describe()
````

````python
df1['Age'].unique()
````

````python
# Replacing unrealistic age values (below 18) with NaN
df1['Age'] = df1['Age'].where(df1['Age'] >= 18, np.nan)

# Filling NaN age values with median age to preserve row count
median_age = df1['Age'].median()
df1['Age'] = df1['Age'].fillna(median_age)

print(f"Median age used for imputation: {median_age}")
print(f"Remaining age anomalies (below 18): {(df1['Age'] < 18).sum()}")
````

````python
# Filling missing Payment_of_Min_Amount values with mode
# 12% missing is too large to drop, mode imputation preserves dataset integrity
mode_payment = df1['Payment_of_Min_Amount'].mode()[0]
df1['Payment_of_Min_Amount'] = df1['Payment_of_Min_Amount'].fillna(mode_payment)

print(f"Mode value used for imputation: {mode_payment}")
print(f"Remaining nulls in Payment_of_Min_Amount: {df1['Payment_of_Min_Amount'].isnull().sum()}")
````

````python
# Final check after all cleaning steps
print("Dataset shape after cleaning:", df1.shape)
print("\nTotal null values per column:")
print(df1.isnull().sum()[df1.isnull().sum() > 0])
print("\nNo null values found." if df1.isnull().sum().sum() == 0 else "")
print("\nAge range after cleaning:")
print(f"Min: {df1['Age'].min()}, Max: {df1['Age'].max()}")
````

````python
df1.dtypes
````

### What all manipulations have you done and insights you found?

A copy of the dataset was created to preserve the original data during analysis. Columns like 'ID','Customer_ID','Name','SSN' were removed because they do not contribute to meaningful analytical insights and only act as identifiers. During data inspection, an irregular value 'NM' was noticed in 'Payment_of_Min_Amount' column, it likely represents missing or unidentified value so it was replaced with 'NaN' for proper missing value handling. After replacement, around 12,007 missing values detected in that column which was earlier looked like valid text.

After that, i inspected numerical column 'Age' for unrealistic values. Although most values in that column were within reasonable range, a few unusually low values like 14,15,16 and 17 spotted and may represent as anomalies or synthetic inconsistencies in the dataset. Unrealistic age values below 18 were identified and replaced with the median age, as individuals under 18 are legally not eligible for credit products. Additionally, the 12,007 missing values in Payment_of_Min_Amount were imputed using the column mode, since dropping 12% of the dataset would significantly reduce analytical reliability. It should be noted that mode imputation introduces an assumption about missing values and may slightly skew the distribution of this column toward the most frequent category.
